```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/cleaned_delivery_data.csv")
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['delivery_time_days'], bins=30, kde=True)
plt.title("Distribution of Delivery Time")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df['delivery_time_days'])
plt.title("Delivery Time Boxplot")
plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    df.select_dtypes(include=['number']).corr(),
    annot=True,
    cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='distance_km',
    y='delivery_time_days',
    data=df)

plt.title("Distance vs Delivery Time")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='purchase_month',
    data=df)

plt.title("Orders by Month")
plt.show()
```
